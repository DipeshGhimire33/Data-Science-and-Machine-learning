## Finding missing value columns

```python
for each in house_df.columns:
    if house_df[each].isnull().sum() > 0 :
        print(each)
```

```python
for each in house_df.columns:
    if house_df[each].isnull().sum() > 0 :
        if house_df[each].dtype == "str":
            mode_value = house_df[each].mode()[0]
            house_df[each] = house_df[each].fillna(mode_value)
        else:
            mean_value = house_df[each].mean()
            house_df[each] = house_df[each].fillna(mean_value)
            
```

```python
for each in house_df.columns:
    if house_df[each].isnull().sum() > 0 :
        print(each)
```

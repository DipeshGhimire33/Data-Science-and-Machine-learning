Numpy Operations

TRIGNOMETRIC OPERATIONS

```python
x = np.array([0,30,45,60,90])

np.sin(x)
np.cos(x)
np.tan(x)

cosec = np.reciprocal(np.sin(x))
cosec
```

```python
sec = np.reciprocal(np.cos(x))
sec
```

```python
cot = np.reciprocal(np.tan(x))
cot
```

Matrix Operation

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([[11,12,13],[14,15,16]])
```

```python
# transpose
bT = np.transpose(b)
```

```python
a @ bT
# np.dot(a, bT)
```

```python
# Trace
d_sum = np.trace(b)
print(d_sum)
```

```python
# Rank
np.linalg.matrix_rank(a)
```

```python
# Determinant

x= np.random.randint(1,5, size= (4,4))
np.linalg.det(x)

# Inverse

invb = np.linalg.inv(x)
invb
```

```python
a = np.array([[4,2],[5,3]])
b = np.array([[8],[9]])

a_inv = np.linalg.inv(a)
np.dot(a_inv, b)
```

```python
solution = np.linalg.solve(a,b)
solution
```

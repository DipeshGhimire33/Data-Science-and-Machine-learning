```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

# 100x100 grayscale image
img = np.random.rand(100, 100)

# Add a simple recognizable smiley
Y, X = np.ogrid[:100, :100]
mask_face = (X - 50)**2 + (Y - 50)**2 < 40**2
img[mask_face] = 0.8

# eyes
img[35:42, 35:42] = 0
img[35:42, 63:70] = 0

# mouth
img[65:70, 35:65] = 0

plt.imshow(img, cmap="gray")
plt.axis("off")
plt.show()
```

Handling Missing Values

```python
data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "John"],
    "Age": [20, 21, None, 22, 20],
    "Marks": [80, None, 75, 90, 85]
}

df = pd.DataFrame(data)
```

```python
df.info()
```

```python
fill_val = df["Age"].mean()
df["Age"] = df["Age"].fillna(fill_val)
df
```

```python
df["Marks"] = df["Marks"].fillna(0)
df
```

```python
df.info()
```

Grouping and Aggregation

```python
data = {
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Salary": [50000, 40000, 60000, 55000, 45000, 65000]
}

df = pd.DataFrame(data)
```

```python
df["Salary"].mean()
```

```python
df["Salary"].max()
```

```python
df["Salary"].min()
```

```python
avg_sal = df.groupby("Department")["Salary"].mean()
print(avg_sal)
```

```python
max_sal =df.groupby("Department")["Salary"].max()
print(max_sal)
```

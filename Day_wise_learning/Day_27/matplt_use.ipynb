Task 8:

Create a figure containing 4 different plots:

* Line plot
* Bar chart
* Scatter plot
* Histogram

Use the following data where appropriate:

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 50]

study_hours = [1, 2, 3, 4, 5, 6]
scores = [50, 55, 65, 70, 80, 90]

marks = [45, 55, 60, 65, 70, 75, 80, 85, 90, 95]
* Arrange the four plots in a 2 × 2 layout.
* Give each plot an appropriate title.


```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 50]

study_hours = [1, 2, 3, 4, 5, 6]
scores = [50, 55, 65, 70, 80, 90]

marks = [45, 55, 60, 65, 70, 75, 80, 85, 90, 95]

fig, ax =plt.subplots(2,2)
ax[0,0].plot(x,y)
ax[0,0].set_title("line plot")

ax[0,1].bar(categories,values)
ax[0,1].set_title("Bar chart")

ax[1,0].scatter(study_hours,scores)
ax[1,0].set_title("Scatter plot")

ax[1,1].hist(marks, bins = 2, edgecolor = "black") 
ax[1,1].set_title("Histogram")

plt.tight_layout()

```

Task 10:

Create the following dataset:

students = ["Alice", "Bob", "Charlie", "David", "Eva"]
math = [85, 78, 92, 88, 95]
science = [80, 82, 90, 85, 93]
english = [88, 75, 85, 90, 91]

Create visualizations to answer the following:

* Create a bar chart of Math marks.
* Create a line plot comparing Math, Science, and English marks.
* Create a scatter plot between Math and Science marks.
* Create a pie chart showing the average marks of the three subjects.
* Add suitable titles, labels, legends, and grid lines where appropriate.

```python
students = ["Alice", "Bob", "Charlie", "David", "Eva"]
math = [85, 78, 92, 88, 95]
science = [80, 82, 90, 85, 93]
english = [88, 75, 85, 90, 91]

fig, ax = plt.subplots(2, 2, figsize=(10, 10))

ax[0,0].bar(students, math)
ax[0,0].set_title("bargrph")

ax[0,1].plot(math ,label = "math")
ax[0,1].plot(english, label = "english")
ax[0,1].plot(science, label = "science")
ax[0,1].grid()
ax[0,1].legend()
ax[0,1].set_title("line plot")

ax[1,0].scatter(math,science)
ax[1,0].scatter(math,science)
ax[1,0].set_xlabel("math")
ax[1,0].set_ylabel("science")

ax[1,0].grid()
ax[1,0].set_title("scatter_plt")



avg_math = np.array(math).mean()
avg_science = np.array(science).mean()
avg_english = np.array(english).mean()

ax[1,1].pie([avg_math,avg_english,avg_science], labels = ["math","english","science"], autopct = "%1.2f%%")
ax[1,1].set_title("pie chart")

plt.tight_layout()
```

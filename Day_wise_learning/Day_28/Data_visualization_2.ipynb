# Heat Map

```python
correl = df.corr(numeric_only= True)
sns.heatmap(correl, cmap = "coolwarm", annot= True,mask = False )
```

```python
import numpy as np

masks = np.triu(np.ones_like(correl , dtype = bool))
```

```python
sns.heatmap(correl, cmap = "coolwarm", annot= True,mask = masks )
```

### Bubble Chart

```python
sns.relplot(
    data = df,
    x ="SibSp",
    y = "Age",
    size = "Fare",
    hue = "Sex",
    alpha = 0.7
    
)
```

# Violin plot

```python
sns.violinplot(
    df,
    x ="Sex",
    y = "Age",
    hue = "Survived",
    inner = "quart",
    split = True
    )
```

### Handling null value for age ,using initials (Mr. ,Mrs, Miss , Master) groups

```python
initials = [name.split(",")[-1].split(".")[0].strip() for name in titanic_df.Name]
```

```python
titanic_df["initials"] = initials
titanic_df.tail()
```

```python
titanic_df["initials"].value_counts()
```

```python
titanic_df.initials = titanic_df.initials.replace(to_replace= ["Rev","Major","Col","Don","Sir","Capt","Jonkheer"],
                            value = "Mr")
titanic_df.initials.value_counts()
```

```python
titanic_df.initials = titanic_df.initials.replace(
    to_replace={"Mlle":"Mrs","Ms":"Mrs","Mme":"Mrs","Lady":"Mrs","the Countess":"Mrs",}
)
```

```python
titanic_df.initials.value_counts()
```

```python
def replace_with_gender(row):
    value = "Mr" if row.Sex == "male" else "Mrs"
    return row.initials.replace("Dr",value , True)

titanic_df.initials = titanic_df.apply(replace_with_gender, axis = 1)
titanic_df.initials.value_counts()
```

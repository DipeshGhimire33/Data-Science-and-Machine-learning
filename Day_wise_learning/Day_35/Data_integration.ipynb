Data integration using join, concat

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "Account No": [1001, 1002, 1003, 1004, 1005],
    "Name": ["Aarav Sharma", "Priya Singh", "Rohan Thapa", "Anisha Gurung",
             "Suman Karki"],
    "Age": [24, 31, 27, 22, 35]
})

df2 = pd.DataFrame({
    "Account No": [2001, 2002, 2003, 2004, 2005],
    "Name": ["Bikash Rai", "Sneha Joshi", "Nabin Shrestha", "Riya Adhikari",
             "Kiran Magar"],
    "Age": [29, 26, 33, 21, 38]
})
```

# "Concatenation"

```python
pd.concat([df1,df2],axis = 0,ignore_index= True)
```

```python
df1 = pd.DataFrame({
    "Account No": [1001, 1002, 1003, 1004, 1005],
    "Name": ["Aarav Sharma", "Priya Singh", "Rohan Thapa", "Anisha Gurung", 
             "Suman Karki"],
    "Age": [24, 31, 27, 22, 35]
})

df2 = pd.DataFrame({
    "Account No": [1001, 1002, 1003, 1004, 1005],
    "principle": [500,600,700,800,900],
    "interest": [2.9, 2.6, 3.3, 2.1, 3.8]
})
```

```python
pd.concat([df1,df2],axis = 1)
```

## Merging

```python
df1 = pd.DataFrame({
    "Account No": [1001, 1002, 1003, 1004, 1005],
    "Name": ["Aarav Sharma", "Priya Singh", "Rohan Thapa", "Anisha Gurung",
             "Suman Karki"],
    "Age": [24, 31, 27, 22, 35]
})

df2 = pd.DataFrame({
    "Account No": [1001, 1002, 1003, 1006, 1007],
    "principle": [500, 600, 700, 800, 900],
    "interest": [2.9, 2.6, 3.3, 2.1, 3.8]
})
```

```python
pd.merge(df1,df2, how = "inner", on = "Account No")
```

```python
pd.merge(df1,df2, how = "outer", on = "Account No")
```

```python
pd.merge(df1,df2, how = "right", on = "Account No")
```

# Join

```python
df1.join(df2, how = "inner" ,lsuffix= "_left")
```

## Machine Learning

Machine learning is a subset of artificial intelligence that uses algorithms to let computers learn from data and make predictions without being explicitly programmed.You can learn more about the core principles and scope of the field by reviewing the GeeksforGeeks Machine Learning Tutorial or reading the Coursera Guide on Machine Learning.

Core Types of Machine Learning
* Supervised learning: Trains a model using labeled data to predict outcomes or classify inputs.
* Unsupervised learning: Analyzes unlabeled data to find hidden structures, clusters, or patterns.
* Reinforcement learning: Guides an agent through trial, error, rewards, and penalties to achieve a goal.
* Semi-supervised learning: Combines a small amount of labeled data with a large pool of unlabeled data

___

**Types of Supervised Learning:**

1. **Classification:** Predicting categorical labels or classes. The output is a discrete label.  
   - Example: Spam detection (spam or not spam), handwriting recognition.

2. **Regression:** Predicting continuous numerical values.  
   - Example: Predicting house prices, stock prices.

---

**Types of Unsupervised Learning:**

1. **Clustering:** Grouping similar data points into clusters based on their features.  
   - Example: Customer segmentation, market segmentation.

2. **Dimensionality Reduction:** Reducing the number of features in the data while preserving important information.  
   - Example: Principal Component Analysis (PCA), t-SNE.

3. **Anomaly Detection:** Identifying unusual data points that do not fit the general pattern.  
   - Example: Fraud detection, network security.

4. **Association Rule Learning:** Discovering interesting relations between variables in large databases.  
   - Example: Market basket analysis (e.g., customers who buy bread also buy butter).

These are common subtypes within supervised and unsupervised learning, tailored to different types of data and tasks.

## Regression

Regression analysis in machine learning is a type of supervised learning used to predict continuous numerical outcomes based on input features. The goal is to model the relationship between the dependent variable (target) and one or more independent variables (features).

**Key Points about Regression in ML:**

- **Purpose:** To estimate the value of a continuous target variable.
- **Examples:** Predicting house prices, stock market trends, temperature forecasting.
- **Models:** Common regression algorithms include Linear Regression, Polynomial Regression, Support Vector Regression, Decision Tree Regression, Random Forest Regression, and Neural Network Regression.

**How it works:**
1. The model learns from historical data with known outcomes.
2. It finds the best-fit function that maps inputs to outputs.
3. Once trained, the model can predict values for new, unseen data.

**Evaluation Metrics:**
- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R-squared (R²)

Regression analysis is fundamental in scenarios where understanding the relationship between variables and predicting continuous outcomes are essential.

## Linear Regression

Linear Regression in Machine Learning is one of the most fundamental and widely used supervised learning algorithms for regression tasks. It models the relationship between a dependent variable (target) and one or more independent variables (features) by fitting a linear equation to observed data.

### How Linear Regression Works:
- **Objective:** To find the best-fit straight line (or hyperplane in multiple dimensions) that minimizes the difference between predicted and actual values.
- **Mathematical Model:**  
  y=β0​+β1​x1​+β2​x2​+...+βn​xn​+ϵ
  where:
* y is the predicted output
* (x_1, x_2, ..., x_n\) are input features
* (\beta_0\) is the intercept
* (\beta_1, \beta_2, ..., \beta_n\) are coefficients (weights)
* (\epsilon\) is the error term

### Key Points:
- **Training:** The algorithm estimates the coefficients (\( \beta \)) by minimizing the sum of squared differences between predicted and actual values (least squares method).
- **Interpretability:** Coefficients indicate the strength and direction of the relationship between features and the target.
- **Assumptions:**
  - Linear relationship between features and target
  - Independence of errors
  - Homoscedasticity (constant variance of errors)
  - Normal distribution of errors (for inference)

### Applications:
- Predicting sales based on advertising spend
- Estimating house prices
- Forecasting sales or revenue

### Advantages:
- Simple and easy to interpret
- Computationally efficient
- Works well when the relationship is approximately linear

### Limitations:
- Not suitable for complex, non-linear relationships
- Sensitive to outliers
- Assumes linearity and other statistical assumptions

Linear regression serves as a foundational technique in machine learning and statistical modeling, providing a baseline for more complex models.

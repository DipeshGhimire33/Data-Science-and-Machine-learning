## Conditional Probability

Conditional probability is the probability of an event **A occurring given that event B has already occurred**.


### Formula

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
$$

Where:

* \(P(A\mid B)\) = probability of **A given B**
* \(P(A\cap B)\) = probability that **both A and B occur**
* \(P(B)\) = probability that **B occurs**, with \(P(B)>0\)

### Example

Suppose a class has **40 students**:

* 24 students study Mathematics.
* 12 students study both Mathematics and Physics.

If a student is known to study Physics, what's the probability they also study Mathematics?

If 20 students study Physics:

$$
P(M\mid P)=\frac{P(M\cap P)}{P(P)}
=\frac{12/40}{20/40}
=\frac{12}{20}
=\boxed{0.6}
$$

So the probability is **60%**.

## Bayes' Theorem

Bayes' theorem is used to find the probability of an event **given new evidence**.

### Formula

$$
P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}
$$

Where:

* \(P(A\mid B)\) = probability of **A given B**
* \(P(B\mid A)\) = probability of **B given A**
* \(P(A)\) = prior probability of A
* \(P(B)\) = probability of the evidence B

## Naive Bayes

Naive Bayes is a classification algorithm based on Bayes' theorem.

It assumes that the features are **conditionally independent given
the class**.

### Formula

$$
P(C\mid X_1,X_2,\ldots,X_n)
=
\frac{P(C)\prod_{i=1}^{n}P(X_i\mid C)}
{P(X_1,X_2,\ldots,X_n)}
$$

For classification:

$$
\hat{C}
=
\arg\max_C
P(C)\prod_{i=1}^{n}P(X_i\mid C)
$$

### Assumption

The naive assumption is:

$$
P(X_1,X_2,\ldots,X_n\mid C)
=
\prod_{i=1}^{n}P(X_i\mid C)
$$

In other words, the features are assumed to be **conditionally
independent given the class**.

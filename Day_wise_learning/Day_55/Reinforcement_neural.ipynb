# Theory of Neural Networks and Reinforcement Learning

## 1. Neural Networks

A **neural network** is a machine-learning model inspired by the structure of biological neurons. It learns a relationship between inputs and outputs by adjusting numerical parameters called **weights**.

### 1.1 Basic Structure

A neural network typically consists of:

* **Input layer** – receives the input features.
* **Hidden layers** – transform the input through mathematical operations.
* **Output layer** – produces the final prediction.

```text
Input Layer       Hidden Layer        Output Layer

 x₁ ───────────►   h₁ ───────────►
 x₂ ───────────►   h₂ ───────────►    ŷ
 x₃ ───────────►   h₃ ───────────►
```

### 1.2 Artificial Neuron

A neuron calculates a weighted sum of its inputs and applies an activation function:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

where:

* $x_i$ = input
* $w_i$ = weight
* $b$ = bias
* $z$ = weighted sum

The neuron then applies an activation function:

$$
a = f(z)
$$

### 1.3 Activation Functions

Activation functions introduce **non-linearity**, allowing neural networks to learn complex relationships.

#### Sigmoid

$$
\sigma(x) = \frac{1}{1+e^{-x}}
$$

Output range:

$$
0 < \sigma(x) < 1
$$

#### ReLU

$$
ReLU(x) = \max(0,x)
$$

ReLU is widely used in hidden layers because it is computationally simple and helps with gradient-based training.

#### Tanh

$$
\tanh(x) = \frac{e^x-e^{-x}}{e^x+e^{-x}}
$$

Its output ranges from $-1$ to $1$.

---

## 2. Forward Propagation

**Forward propagation** is the process of passing input data through the network to obtain a prediction.

For a simple layer:

$$
z = Wx+b
$$

Then:

$$
a = f(z)
$$

The output of one layer becomes the input to the next layer.

```text
Input
  ↓
Weighted Sum
  ↓
Activation Function
  ↓
Hidden Layer
  ↓
Output
```

---

## 3. Loss Function

The network needs a way to measure how different its prediction is from the correct answer.

This is done using a **loss function**.

For mean squared error:

$$
L = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

where:

* $y_i$ = actual value
* $\hat{y}_i$ = predicted value

The objective of training is generally to **minimize the loss**.

---

## 4. Backpropagation

**Backpropagation** calculates how much each weight contributed to the error.

It uses the **chain rule of calculus** to calculate gradients:

$$
\frac{\partial L}{\partial w}
$$

The weights are then updated using an optimization method such as gradient descent.

### Gradient Descent

$$
w_{new}=w_{old}-\eta\frac{\partial L}{\partial w}
$$

where $\eta$ is the **learning rate**.

```text
Prediction
    ↓
Calculate Loss
    ↓
Backpropagation
    ↓
Calculate Gradients
    ↓
Update Weights
    ↓
Repeat
```

# 5. Reinforcement Learning

**Reinforcement Learning (RL)** is a type of machine learning in which an **agent learns to make decisions by interacting with an environment**.

Instead of being given the correct answer for every input, the agent receives **rewards or penalties** based on its actions.

### 5.1 Main Components

| Component       | Meaning                           |
| --------------- | --------------------------------- |
| **Agent**       | Learner or decision maker         |
| **Environment** | World in which the agent operates |
| **State**       | Current situation                 |
| **Action**      | Choice made by the agent          |
| **Reward**      | Feedback received after an action |
| **Policy**      | Strategy used to select actions   |

The basic interaction is:

```text
        Action
Agent ───────────► Environment
  ▲                    │
  │                    │
  └──── State + Reward ┘
```

# 6. Markov Decision Process

Many reinforcement-learning problems are modeled as a **Markov Decision Process (MDP)**.

An MDP can be represented as:

$$
(S,A,P,R,\gamma)
$$

where:

* $S$ = set of states
* $A$ = set of actions
* $P$ = transition probability
* $R$ = reward function
* $\gamma$ = discount factor

The **Markov property** means that the future depends on the current state and action rather than the complete history.

---

# 7. Policy

A **policy** determines which action the agent should take.

It can be written as:

$$
\pi(a|s)
$$

which represents the probability of taking action $a$ when the agent is in state $s$.

A deterministic policy can be written as:

$$
a = \pi(s)
$$

---

# 8. Reward and Return

The immediate reward is represented by:

$$
R_{t+1}
$$

The agent generally tries to maximize its **expected cumulative discounted reward**.

The return is:

$$
G_t = R_{t+1}+\gamma R_{t+2}+\gamma^2R_{t+3}+\cdots
$$

where:

$$
0\leq\gamma\leq1
$$

A larger $\gamma$ gives more importance to future rewards.

---

# 9. Value Function

The **state-value function** measures the expected return from a state under a policy:

$$
V^\pi(s)=E_\pi[G_t|S_t=s]
$$

The **action-value function**, or **Q-function**, measures the expected return after taking action $a$ in state $s$:

$$
Q^\pi(s,a)=E_\pi[G_t|S_t=s,A_t=a]
$$

---

# 10. Bellman Equation

The Bellman equation expresses a value in terms of the immediate reward and future value.

For the state-value function:

$$
V^\pi(s)
=
\sum_a\pi(a|s)
\sum_{s'}P(s'|s,a)
\left[
R(s,a,s')+\gamma V^\pi(s')
\right]
$$

For Q-learning, the optimal Bellman equation is:

$$
Q^*(s,a)
=
E
\left[
R+\gamma\max_{a'}Q^*(s',a')
\right]
$$

---

# 11. Q-Learning

**Q-learning** is an off-policy reinforcement-learning algorithm.

Its update rule is:

$$
Q(s,a)
\leftarrow
Q(s,a)
+
\alpha
\left[
r+\gamma\max_{a'}Q(s',a')
-
Q(s,a)
\right]
$$

where:

* $\alpha$ = learning rate
* $\gamma$ = discount factor
* $r$ = reward
* $s'$ = next state
* $a'$ = possible next action

The term inside the brackets is the **temporal-difference error**.

---

# 12. Exploration vs. Exploitation

An RL agent faces two choices:

### Exploration

Try new actions to discover potentially better strategies.

### Exploitation

Choose the action currently believed to be best.

A common approach is **ε-greedy exploration**:

* With probability $\epsilon$: choose a random action.
* With probability $1-\epsilon$: choose the action with the highest estimated value.

```text
             Agent
               │
        ┌──────┴──────┐
        │             │
   Explore         Exploit
   random          best-known
    action           action
```

---

# 13. Deep Reinforcement Learning

When a neural network is used to approximate value functions or policies, the approach is called **Deep Reinforcement Learning (Deep RL)**.

For example, **Deep Q-Network (DQN)** uses a neural network to approximate:

$$
Q(s,a;\theta)
$$

where $\theta$ represents the neural-network parameters.

Instead of storing a Q-value for every state-action pair, the neural network learns a general approximation.

```text
State
  │
  ▼
Neural Network
  │
  ├── Q(action 1)
  ├── Q(action 2)
  ├── Q(action 3)
  └── Q(action 4)
          │
          ▼
    Select Action
```

---

# 14. Neural Networks vs. Reinforcement Learning

| Feature            | Neural Network                     | Reinforcement Learning             |
| ------------------ | ---------------------------------- | ---------------------------------- |
| Main idea          | Learn patterns/functions           | Learn actions through rewards      |
| Learning signal    | Usually labeled data or loss       | Rewards and environment feedback   |
| Main components    | Layers, weights, activations       | Agent, state, action, reward       |
| Goal               | Minimize prediction loss           | Maximize cumulative reward         |
| Example            | Image classification               | Game-playing agent                 |
| Typical algorithms | Backpropagation + gradient descent | Q-learning, SARSA, policy gradient |

Importantly, **neural networks and reinforcement learning are not competing concepts**. A neural network can be used as the function approximator inside a reinforcement-learning system.

---

# 15. Relationship Between Neural Networks and RL

A modern RL system can combine both ideas:

$$
\text{Environment}
\rightarrow
\text{State}
\rightarrow
\text{Neural Network}
\rightarrow
\text{Action}
\rightarrow
\text{Reward}
\rightarrow
\text{Learning}
$$

For example, a neural network can learn a policy:

$$
\pi_\theta(a|s)
$$

where $\theta$ represents the neural-network parameters.

The parameters are updated to improve the expected cumulative reward.

# 16. Applications

### Neural Networks

* Image recognition
* Speech recognition
* Natural language processing
* Medical image analysis
* Time-series prediction
* Recommendation systems

### Reinforcement Learning

* Robotics
* Game playing
* Autonomous systems
* Resource allocation
* Control systems
* Sequential decision making

### Deep Reinforcement Learning

Combines neural networks with RL and is useful when the state space is large or complex, such as images, sensor data, or high-dimensional control problems.

---

# 17. Summary

**Neural Networks:**

> Learn complex mappings from inputs to outputs by adjusting weights through optimization.

**Reinforcement Learning:**

> Learns how to make sequential decisions by interacting with an environment and maximizing cumulative reward.

**Deep Reinforcement Learning:**

> Uses neural networks to represent policies, value functions, or Q-functions in reinforcement-learning problems.

The central learning concepts can be summarized as:

$$
\boxed{\text{Neural Network} \rightarrow \text{Prediction} \rightarrow \text{Loss} \rightarrow \text{Gradient} \rightarrow \text{Weight Update}}
$$

and

$$
\boxed{\text{RL} \rightarrow \text{State} \rightarrow \text{Action} \rightarrow \text{Reward} \rightarrow \text{Policy/Value Update}}
$$

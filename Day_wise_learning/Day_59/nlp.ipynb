# Introduction to NLP in Machine Learning

## What is NLP?

**Natural Language Processing (NLP)** is a branch of **Artificial Intelligence (AI)** and **Machine Learning (ML)** that enables computers to understand, interpret, process, and generate human language.

Human language can be in the form of **text or speech**. NLP helps machines work with languages such as English, Nepali, Hindi, and many others.

### Examples of NLP

* **Chatbots** — understand and respond to user questions.
* **Machine Translation** — translates text from one language to another.
* **Sentiment Analysis** — determines whether a text expresses positive, negative, or neutral feelings.
* **Spam Detection** — identifies unwanted emails or messages.
* **Speech Recognition** — converts spoken language into text.
* **Text Summarization** — creates shorter summaries of long documents.
* **Question Answering** — provides answers to questions written in natural language.

## NLP and Machine Learning

Machine Learning allows computers to learn patterns from data instead of being explicitly programmed for every task. In NLP, machine learning algorithms are trained using large collections of text called **corpora**.

A typical NLP system follows these steps:

```text
Raw Text
   ↓
Text Preprocessing
   ↓
Feature Extraction / Text Representation
   ↓
Machine Learning Model
   ↓
Prediction / Output
```

## Common NLP Techniques

### 1. Tokenization

Tokenization divides text into smaller units called **tokens**, such as words or sentences.

Example:

```text
"I love machine learning."
```

can be tokenized as:

```text
["I", "love", "machine", "learning"]
```

### 2. Stop Word Removal

Stop words are common words that may provide little useful information for some NLP tasks.

Examples:

```text
the, is, a, an, of, and
```

### 3. Stemming

Stemming reduces words to their basic or root-like form.

```text
playing → play
played  → play
plays   → play
```

### 4. Lemmatization

Lemmatization converts words into their meaningful dictionary form.

```text
better → good
running → run
```

### 5. Part-of-Speech Tagging

POS tagging identifies the grammatical role of each word.

```text
The/DET cat/NOUN sleeps/VERB.
```

## Text Representation

Machine learning models work with numerical data, so text must be converted into numbers.

Common approaches include:

* **Bag of Words (BoW)**
* **TF-IDF**
* **Word Embeddings**
* **Word2Vec**
* **GloVe**
* **Transformer-based embeddings**

For example, the sentence:

```text
"I like NLP"
```

can be represented as a numerical vector that a machine learning model can process.

| Sentiment Analysis  | Analyzing opinions            |

| Spam Detection      | Detecting unwanted messages   |

## NLP with Modern Machine Learning

Modern NLP heavily uses **Deep Learning**, particularly **Transformer architectures**. Models such as **BERT, GPT, and T5** can learn complex relationships between words and sentences.

These models have significantly improved tasks such as:

* Text generation
* Translation
* Question answering
* Text classification
* Information extraction
* Conversational AI

## Conclusion

**Natural Language Processing (NLP)** combines linguistics, artificial intelligence, and machine learning to enable computers to work with human language. From simple techniques such as tokenization and TF-IDF to advanced Transformer models, NLP is an important area of modern machine learning and AI.

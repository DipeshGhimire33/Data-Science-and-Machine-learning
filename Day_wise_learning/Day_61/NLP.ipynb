 # NLP Theory: From Frequency to Context in Machine Learning

 ## 1\. Introduction

 **Natural Language Processing (NLP)** is a field of Artificial Intelligence (AI) and Machine Learning (ML) concerned with enabling computers to process, understand, represent, and generate human language.

 The history of NLP can be understood as a gradual evolution in how machines represent language:

```
Rule-Based NLP
      ↓
Frequency-Based NLP
      ↓
Word2Vec / Word Embeddings
      ↓
Contextual Representations
      ↓
Attention
      ↓
Transformers
      ↓
Large Language Models
```

 The most important conceptual transition is:

 > **Frequency → Representation → Context**

 Early NLP focused on **counting words**. Word2Vec introduced the idea of **learning meaningful vector representations from word co-occurrence**. Later models introduced genuinely **context-dependent representations**, allowing the same word to have different representations depending on its surrounding text.

---

 # 2\. Early NLP: Rule-Based Approaches

 Before statistical machine learning became dominant, many NLP systems were based on **manually written linguistic rules**.

 A rule-based system might contain:

 - Dictionaries
- Grammar rules
- Morphological rules
- Syntactic rules
- Hand-crafted patterns
- Expert knowledge

 For example:

```
Sentence → Noun Phrase + Verb Phrase

Noun Phrase → Determiner + Noun

Verb Phrase → Verb + Noun Phrase
```

 A system could use these rules to analyze sentences.

 ### Advantages

 - Rules can be interpretable.
- Useful for highly constrained domains.
- Can encode explicit linguistic knowledge.

 ### Limitations

 Natural language is extremely complex and ambiguous.

 For example:

 > "I saw the man with the telescope."

 This could mean:

 1. I used a telescope to see the man.
2. I saw a man who had a telescope.

 Writing rules to handle every such possibility becomes extremely difficult.

 This motivated the move toward **data-driven statistical NLP**.

---

 # 3\. Frequency-Based NLP

 A major shift occurred when researchers began treating language as **data that could be counted and modeled statistically**.

 The central idea was:

 > **The patterns that occur frequently in language can provide useful information about how language works.**

 Instead of manually specifying every linguistic rule, a model could learn patterns from a large corpus.

---

 ## 3.1 Word Frequency

 The simplest statistical approach is to count how frequently words occur.

 The probability of a word can be approximated by:

 $$
P(w)=\frac{\text{count}(w)}{\text{total number of words}}
$$

 For example, if `the` occurs 10,000 times in a corpus containing 100,000 words:

 $$
P(\text{the})=\frac{10000}{100000}=0.1
$$

 This provides useful statistical information about language.

 However, frequency alone does not tell us much about **meaning**.

 For example:

```
cat → occurs frequently
dog → occurs frequently
```

 Knowing their frequencies does not tell a machine that `cat` and `dog` are semantically related.

---

 # 4\. N-Gram Models

 Frequency-based NLP developed into **n-gram language models**.

 An n-gram is a sequence of $n$ words.

 Examples:

```
Unigram:
the

Bigram:
machine learning

Trigram:
natural language processing
```

 A bigram model approximates the probability of a sentence as:

 $$
P(w_1,w_2,\ldots,w_n)
\approx
\prod_i P(w_i|w_{i-1})
$$

 For example:

```
I am going home
```

 A bigram model considers:

```
I → am
am → going
going → home
```

 A trigram model considers a larger local context:

```
I am → going
am going → home
```

 ### Advantages of N-Grams

 - Simple to understand.
- Easy to train.
- Useful for early speech recognition.
- Useful for spelling correction.
- Useful for statistical machine translation.
- Provided an important foundation for statistical NLP.

 ### Limitations

 N-gram models have a **fixed context window**.

 They also suffer from:

 - Data sparsity
- Large vocabulary requirements
- Poor handling of long-distance dependencies
- Limited semantic understanding

 For example:

 > "The animal that was sitting near the door suddenly ran away because it was frightened."

 Understanding what **"it"** refers to may require information much earlier in the sentence than a small n-gram window can provide.

---

 # 5\. TF-IDF and the Bag-of-Words Representation

 Another important frequency-based representation was **Bag-of-Words (BoW)**.

 The basic idea is to represent a document according to the words it contains while largely ignoring word order.

 For example:

```
"The cat eats fish"
```

 could be represented as:

```
cat: 1
eats: 1
fish: 1
the: 1
```

 This is useful for tasks such as:

 - Document classification
- Spam detection
- Information retrieval
- Search
- Text clustering

 However, Bag-of-Words has a major weakness:

 > **It does not understand semantic relationships between words.**

 For example:

```
car
automobile
vehicle
```

 are treated as separate features.

---

 ## 5.1 TF-IDF

 **Term Frequency–Inverse Document Frequency (TF-IDF)** improved upon simple word counts by estimating how important a word is to a document.

 A simplified formula is:

 $$
TFIDF(t,d)=TF(t,d)\times IDF(t)
$$

 where:

 - $TF(t,d)$ = frequency of term $t$ in document $d$
- $IDF(t)$ = inverse document frequency

 A word that occurs frequently in one document but rarely across the entire collection receives a higher importance score.

 TF-IDF became widely useful for:

 - Search engines
- Information retrieval
- Document classification
- Text similarity

 However, it still represented words primarily as independent dimensions.

 This led to one of the biggest improvements in NLP:

 > **Word embeddings.**

---

 # 6\. Word2Vec: The Major Bridge from Frequency to Context

 **Word2Vec**, introduced by Mikolov and colleagues in 2013, was a major breakthrough in NLP.

 It is especially important historically because it sits **between traditional frequency-based NLP and modern contextual NLP**.

 Word2Vec did not simply count words. Instead, it learned **dense vector representations of words from their surrounding contexts**.

 The key theoretical idea comes from the **distributional hypothesis**:

 > **Words that occur in similar contexts tend to have similar meanings.**

---

 ## 6.1 From Counts to Vectors

 Traditional frequency-based NLP might represent words using counts:

```
cat = frequency 1520
dog = frequency 1390
```

 Word2Vec instead learns a vector for each word:

```
cat → [0.21, -0.45, 0.73, 0.18, ...]
dog → [0.19, -0.42, 0.69, 0.21, ...]
car → [-0.71, 0.34, -0.12, 0.58, ...]
```

 These vectors are called **word embeddings**.

 The important point is that the vector is learned from the statistical patterns in the corpus.

---

 # 7\. How Word2Vec Learns Meaning

 Word2Vec uses the surrounding words of a target word to learn its representation.

 Consider:

```
The cat drinks milk.
The dog drinks water.
The cat plays outside.
The dog plays outside.
```

 The words `cat` and `dog` occur in similar contexts:

```
The ___ drinks
The ___ plays
```

 Therefore, Word2Vec learns vectors for `cat` and `dog` that are close to each other in the embedding space.

 Conceptually:

```
             dog
              ●
             /
            /
           ● cat

                     ● car
```

 The distance between vectors can represent aspects of semantic similarity.

---

 # 8\. Word2Vec's Immense Improvement

 Word2Vec represented a fundamental shift:

```
Frequency-Based NLP
        ↓
Count occurrences
        ↓
Observe co-occurrence
        ↓
Learn statistical relationships
        ↓
Word2Vec
        ↓
Convert relationships into dense vectors
```

 The question changed from:

 > **"How often does this word occur?"**

 to:

 > **"What words occur around this word, and what does that tell us about its meaning?"**

 This was an enormous improvement because the model could now learn **semantic relationships**.

 For example, embeddings could capture relationships such as:

 $$
king - man + woman \approx queen
$$

 The exact quality of such analogies depends on the model and training data, but the example illustrates the basic idea: semantic relationships can be represented geometrically.

---

 # 9\. Word2Vec Models: CBOW and Skip-Gram

 Word2Vec was commonly implemented through two main architectures.

 ## 9.1 Continuous Bag-of-Words (CBOW)

 CBOW predicts a target word from surrounding context words.

 For example:

```
The cat ___ milk
```

 The model uses surrounding words to predict:

```
drinks
```

 Conceptually:

```
context words
      ↓
[The, cat, milk]
      ↓
     model
      ↓
    drinks
```

---

 ## 9.2 Skip-Gram

 Skip-Gram does the opposite.

 It uses a target word to predict surrounding context words.

 For example:

```
Target:
cat

Context:
the
drinks
milk
```

 Conceptually:

```
       cat
        ↓
 ┌──────┼──────┐
 ↓      ↓      ↓
the   drinks   milk
```

 Both approaches learn useful word representations from the statistical structure of language.

---

 # 10\. Why Word2Vec Was Not Yet Fully Contextual

 Despite its enormous improvement, Word2Vec had an important limitation:

 > **A word generally has one learned vector regardless of the sentence in which it appears.**

 Consider:

```
I deposited money in the bank.

I sat beside the river bank.
```

 The meaning of `bank` is different in the two sentences.

 But a traditional Word2Vec model essentially learns:

```
bank → one vector
```

 It does not dynamically produce:

```
bank + financial context → financial-bank representation

bank + river context → river-bank representation
```

 Therefore, Word2Vec was **context-informed but not context-dependent**.

 This distinction is extremely important.

---

 # 11\. Word2Vec as the Bridge

 The historical importance of Word2Vec can therefore be summarized as:

```
Frequency
   ↓
Word co-occurrence
   ↓
Word2Vec
   ↓
Dense semantic representations
   ↓
Contextual representations
```

 Word2Vec still relied fundamentally on **frequency and co-occurrence statistics**, but it transformed those statistics into a much richer representation.

 Therefore:

 > **Frequency told NLP how often words occur. Word2Vec learned what words are related based on their contexts. Contextual models later learned that the representation of a word itself should change according to its current context.**

 This makes Word2Vec one of the most important intermediate steps in the history of NLP.

---

 # 12\. Distributional Hypothesis

 The theoretical foundation behind Word2Vec is the **distributional hypothesis**.

 The idea is often summarized as:

 > **You shall know a word by the company it keeps.**

 In other words, the surrounding linguistic environment provides information about a word's meaning.

 For example:

```
The cat drinks milk.
The dog drinks water.
```

 Because `cat` and `dog` appear in similar linguistic environments, we can infer that they have related meanings.

 This principle became one of the foundations of modern representation learning in NLP.

---

 # 13\. From Static Embeddings to Contextual Representations

 Word2Vec demonstrated that words could be represented meaningfully as vectors.

 The next challenge was:

 > **Can the representation of a word change depending on its context?**

 Consider:

```
The bank approved my loan.

The river bank was flooded.
```

 A modern contextual model should produce different representations for `bank` in the two sentences.

 Conceptually:

```
bank + loan + approved
          ↓
financial meaning

bank + river + flooded
          ↓
geographical meaning
```

 This led to **contextual word representations**.

---

 # 14\. RNNs: Sequential Context

 **Recurrent Neural Networks (RNNs)** introduced a mechanism for processing language sequentially while maintaining a hidden state.

 A simplified RNN can be represented as:

 $$
h_t=f(x_t,h_{t-1})
$$

 where:

 - $x_t$ = current input
- $h_{t-1}$ = previous hidden state
- $h_t$ = current hidden state

 For:

```
The → cat → sat → on → the → mat
```

 the model processes the sequence step by step.

 The hidden state carries information from earlier words.

---

 # 15\. LSTM: Better Long-Term Dependencies

 Standard RNNs can have difficulty preserving information over long sequences.

 **Long Short-Term Memory (LSTM)** networks introduced gates that help control what information should be:

 - Remembered
- Forgotten
- Updated
- Passed forward

 LSTMs became important in:

 - Machine translation
- Speech recognition
- Text generation
- Sentiment analysis
- Sequence classification

 However, RNNs and LSTMs still process sequences sequentially, which makes training less parallelizable and makes very long-range relationships difficult.

---

 # 16\. Attention: A Major Shift Toward Context

 The next major breakthrough was the **attention mechanism**.

 Instead of forcing the model to rely primarily on a single sequential hidden state, attention allows the model to examine different parts of the input and determine which ones are important.

 Consider:

 > "The animal didn't cross the road because it was tired."

 To interpret `it`, the model needs to identify the relevant earlier information.

 Attention provides a mechanism for the model to assign different weights to different tokens.

 Conceptually:

```
The animal didn't cross the road because it was tired.
    ↑                                      ↑
    └──────── relevant relationship ───────┘
```

 The central idea is:

 > **Not all parts of the context are equally important.**

---

 # 17\. Transformers

 The **Transformer architecture** represented another fundamental change in NLP.

 Transformers rely heavily on **self-attention** rather than recurrent processing.

 The basic attention operation can be written as:

 $$
Attention(Q,K,V)
=
softmax\left(\frac{QK^T}{\sqrt{d_k}}\right)V
$$

 where:

 - $Q$ = queries
- $K$ = keys
- $V$ = values
- $d_k$ = key dimensionality

 Self-attention allows tokens to directly interact with other tokens in the sequence.

 For example:

```
The animal didn't cross the road because it was tired.
```

 The representation of `it` can directly consider relevant words elsewhere in the sentence.

---

 # 18\. From Transformers to Large Language Models

 Transformers made it practical to train very large language models on massive amounts of text.

 Important transformer-based model families include:

 - BERT
- GPT
- T5
- RoBERTa

 These models can learn patterns involving:

 - Syntax
- Semantics
- Word relationships
- Long-range dependencies
- Context
- Language generation
- Knowledge encoded in training data

 The scale of training also became an important factor.

 The progression became:

```
Word
  ↓
Word + Local Context
  ↓
Sentence
  ↓
Document
  ↓
Large Context
  ↓
Large Language Model
```

---

 # 19\. The Complete Evolution

 The historical development can now be represented more precisely:

```
RULE-BASED NLP
      │
      │ Hand-written linguistic rules
      ↓
FREQUENCY-BASED NLP
      │
      │ Count words and word sequences
      ↓
N-GRAMS / TF-IDF
      │
      │ Learn statistical co-occurrence patterns
      ↓
WORD2VEC / WORD EMBEDDINGS
      │
      │ Convert co-occurrence statistics
      │ into dense semantic vectors
      ↓
CONTEXTUAL REPRESENTATIONS
      │
      │ Same word can have different
      │ representations in different contexts
      ↓
RNN / LSTM
      │
      │ Model sequential dependencies
      ↓
ATTENTION
      │
      │ Dynamically focus on relevant information
      ↓
TRANSFORMERS
      │
      │ Large-scale self-attention
      ↓
LARGE LANGUAGE MODELS
      │
      │ Learn language patterns at massive scale
      ↓
MODERN GENERATIVE NLP
```

 # 20\. Frequency vs. Word2Vec vs. Context

 | Stage | Representation | Main Idea | Context |
| --- | --- | --- | --- |
| Word Frequency | Counts | How often does a word occur? | None |
| Bag-of-Words | Sparse counts | Which words occur in a document? | Very limited |
| N-Grams | Word sequences | Which words occur together locally? | Fixed window |
| TF-IDF | Weighted counts | Which words are important? | Document-level |
| **Word2Vec** | Dense vectors | Which words occur in similar contexts? | Context-informed |
| GloVe / FastText | Dense vectors | Learn semantic relationships | Mostly static |
| RNN | Hidden states | What came before? | Sequential |
| LSTM | Memory states | What information should be retained? | Longer sequential context |
| Attention | Weighted relationships | Which context is relevant? | Dynamic |
| Transformer | Self-attention | How are tokens related? | Broad context |
| LLM | Large contextual representations | How can language patterns be modeled and generated? | Large-scale context |

 # 21\. The Three Major Stages

 The evolution can be simplified into three particularly important stages:

 ## Stage 1: Frequency

 The machine asks:

 > **How often does this word or phrase occur?**

 Examples:

 - Word counts
- Bag-of-Words
- N-grams
- TF-IDF

 Representation:

```
word → count
```

---

 ## Stage 2: Word2Vec and Embeddings

 The machine asks:

 > **What words occur in similar contexts?**

 Representation:

```
word → vector
```

 For example:

```
king → [ ... ]
queen → [ ... ]
man → [ ... ]
woman → [ ... ]
```

 The model can learn semantic relationships from word co-occurrence.

 This is the crucial bridge between frequency and contextual NLP.

---

 ## Stage 3: Context

 The machine asks:

 > **What does this word mean in this particular context?**

 Representation:

```
word + context → contextual vector
```

 For example:

```
bank + money + loan
        ↓
financial meaning

bank + river + water
        ↓
river-bank meaning
```

 This is the foundation of modern contextual NLP.

---

 # 22\. Why the Transition Matters

 The evolution of NLP is not simply a sequence of newer algorithms. It represents a change in the **theory of language representation**.

 ### Frequency-based view

 Language can be modeled through:

 $$
\text{counts} + \text{probabilities}
$$

 ### Word2Vec view

 Language can be represented through:

 $$
\text{co-occurrence} \rightarrow \text{semantic vectors}
$$

 ### Contextual view

 Language can be represented through:

 $$
\text{word}+\text{context}
\rightarrow
\text{meaning}
$$

 ### Transformer view

 Language can be represented through:

 $$
\text{tokens}
+
\text{self-attention}
+
\text{large-scale training}
\rightarrow
\text{contextual representations}
$$

---

 # 23\. A Simple Example of the Entire Evolution

 Consider the sentence:

 > **"The bank approved the loan."**

 ### Frequency-Based Model

 The model sees:

```
bank → frequency
approved → frequency
loan → frequency
```

 It knows how often these words occur.

 ### N-Gram Model

 It learns patterns such as:

```
bank → approved
approved → the
the → loan
```

 It captures local word-sequence probabilities.

 ### Word2Vec

 It learns that:

```
bank
loan
money
credit
finance
```

 occur in related contexts.

 Therefore, their vectors may be relatively close in semantic space.

 ### Contextual Model

 It sees:

```
bank + approved + loan
```

 and determines that `bank` refers to a **financial institution**.

 If the sentence were:

 > **"The bank was flooded after the river overflowed."**

 the contextual representation of `bank` would instead reflect the **side of a river**.

 ### Transformer

 Self-attention allows the representation of `bank` to interact with relevant tokens such as:

```
approved
loan
```

 or:

```
river
flooded
```

 depending on the sentence.

 The history of NLP can therefore be summarized as a movement through increasingly powerful representations:

```
COUNTING
   ↓
"How often does it occur?"

      ↓

CO-OCCURRENCE
   ↓
"What words occur around it?"

      ↓

WORD2VEC
   ↓
"What semantic relationships can be learned
 from those co-occurrence patterns?"

      ↓

CONTEXT
   ↓
"What does this word mean here?"

      ↓

ATTENTION
   ↓
"Which parts of the context are relevant?"

      ↓

TRANSFORMERS
   ↓
"How are all the tokens related?"

      ↓

LLMs
   ↓
"What language patterns can be learned
and used for understanding and generation?"
```

 # 25\. Conclusion

 The history of NLP is fundamentally a history of **better representations of language**.

 Early NLP relied heavily on **rules**. Statistical NLP then shifted toward **frequency, probability, and word co-occurrence**. N-grams and TF-IDF made language computationally measurable, but they remained limited in their ability to represent meaning.

 **Word2Vec was a crucial intermediate breakthrough.** It took statistical information about word co-occurrence and transformed it into **dense semantic vector representations**. This was an immense improvement over simple frequency-based representations because words were no longer treated merely as independent counts. Their relationships and semantic similarities could be represented geometrically.

 However, Word2Vec still assigned essentially one vector to a word. It was therefore **context-informed but not fully context-dependent**.

 RNNs and LSTMs introduced sequential context. Attention provided a mechanism for dynamically focusing on relevant parts of the input. Transformers then made large-scale self-attention practical, leading to today's large language models.

 The overall evolution can therefore be summarized as:

 > **Frequency → Co-occurrence → Word2Vec → Representation → Context → Attention → Transformers → Large Language Models**

 Or, even more simply:

 > **From counting words, to representing words, to understanding words through context.**

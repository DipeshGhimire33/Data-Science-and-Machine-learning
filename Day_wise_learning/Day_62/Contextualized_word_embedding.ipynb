# Contextualized Word Embeddings and How BERT Works

 ## 1\. Contextualized Word Embeddings

 **Contextualized word embeddings** are vector representations whose meaning changes according to the surrounding context.

 This was a major improvement over **static word embeddings** such as Word2Vec, GloVe, and FastText.

 The key difference is:

 $$
\text{Static: } word \rightarrow vector
$$

 $$
\text{Contextual: } word + context \rightarrow vector
$$

 For example:

```
I deposited money in the bank.
```

 and:

```
I sat beside the bank of the river.
```

 The word `bank` has different meanings. A contextual model can therefore generate different representations for `bank` in the two sentences.

---

 # 2\. From Word2Vec to Contextualized Embeddings

 The historical progression is:

```
Frequency
    ↓
N-Grams / TF-IDF
    ↓
Word2Vec / GloVe
    ↓
Static Word Embeddings
    ↓
Contextualized Word Embeddings
    ↓
Attention
    ↓
Transformers
    ↓
BERT / GPT / Other Large Language Models
```

 The important transition is:

 > **Word2Vec learned a general representation of a word from its surrounding words. Contextualized embeddings learned a representation of a word that changes according to its specific context.**

---

 # 3\. Why Word2Vec Was Not Enough

 Word2Vec was a major improvement because it represented words as dense vectors.

 For example:

```
cat → [0.21, -0.45, 0.73, ...]
dog → [0.19, -0.42, 0.69, ...]
```

 Because `cat` and `dog` occur in similar contexts, their vectors can be similar.

 However, Word2Vec essentially gives:

```
bank → one vector
```

 regardless of whether `bank` means:

```
financial institution
```

 or:

```
side of a river
```

 Therefore:

 > **Word2Vec is context-informed, but not fully context-dependent.**

---

 # 4\. What Does Contextualization Mean?

 Suppose we have:

```
The bank approved my loan.
```

 The representation of `bank` can be influenced by:

```
approved
loan
```

 Now consider:

```
The river bank was flooded.
```

 The representation of `bank` can be influenced by:

```
river
flooded
```

 Therefore:

```
bank + financial context
        ↓
financial representation
```

 while:

```
bank + river context
        ↓
geographical representation
```

 Mathematically:

 $$
h_i = f(x_i, x_1,x_2,\ldots,x_n)
$$

 where $h_i$ is the contextual representation of token $x_i$.

---

 # 5\. Attention: The Key to Contextualization

 A major breakthrough was the **attention mechanism**.

 Attention allows a model to determine which other words are relevant to the current word.

 Consider:

 > **The animal didn't cross the road because it was tired.**

 To understand `it`, the model needs to look at other words in the sentence.

 Attention allows the representation of one token to incorporate information from other tokens.

 Conceptually:

```
The animal didn't cross the road because it was tired.
     ↑                                      ↑
     └──────── relevant relationship ───────┘
```

 This idea became central to the **Transformer architecture**.

---

 # 6\. What Is BERT?

 **BERT** stands for:

 > **Bidirectional Encoder Representations from Transformers**

 BERT was introduced by Google researchers in 2018 and became one of the most influential models in NLP.

 BERT is important because it uses the **Transformer encoder** to create deeply contextualized representations.

 Its central idea is:

 > **Understand each word by looking at the words around it, both before and after it.**

---

 # 7\. Why "Bidirectional"?

 Earlier language models often processed text primarily in one direction.

 For example:

```
The → bank → approved → the → loan
```

 A left-to-right model predicts the next word using previous words.

 BERT instead uses information from **both directions**.

 For:

```
The bank approved the loan.
```

 the representation of `bank` can use:

```
Left context:
The

Right context:
approved the loan
```

 So:

```
         left context
              ↓
The → [BANK] ← approved the loan
              ↑
         right context
```

 This gives BERT a much richer understanding of the word.

---

 # 8\. BERT's Architecture

 BERT is based on the **Transformer encoder**.

 A simplified BERT pipeline is:

```
Raw Text
   ↓
Tokenization
   ↓
Token + Position + Segment Embeddings
   ↓
Transformer Encoder
   ↓
Self-Attention
   ↓
Feed-Forward Network
   ↓
Repeated Transformer Layers
   ↓
Contextualized Token Representations
   ↓
NLP Task
```

---

 # 9\. Step 1: Tokenization

 BERT does not directly process raw words.

 The input sentence is first converted into **tokens**.

 For example:

```
"I love natural language processing."
```

 might be represented approximately as:

```
[CLS] I love natural language processing . [SEP]
```

 BERT uses **WordPiece tokenization**, so uncommon words can also be divided into smaller subword units.

 For example, a word may be represented as:

```
unbelievable
      ↓
un + ##believ + ##able
```

 The exact tokenization depends on the vocabulary.

---

 # 10\. Step 2: Special Tokens

 BERT uses special tokens.

 ### `[CLS]`

 Placed at the beginning of the input.

```
[CLS] The bank approved the loan. [SEP]
```

 The final representation of `[CLS]` can be used for sentence-level tasks such as classification.

 ### `[SEP]`

 Marks the end of a sentence or separates two input sequences.

 For example:

```
[CLS] Sentence A [SEP] Sentence B [SEP]
```

---

 # 11\. Step 3: Input Embeddings

 BERT creates the initial representation of each token using three types of information:

```
Token Embedding
       +
Position Embedding
       +
Segment Embedding
       ↓
Input Representation
```

 ### Token Embedding

 Represents the identity of the token.

 ### Position Embedding

 Provides information about where the token occurs in the sequence.

 This is important because self-attention by itself does not inherently know token order.

 For example:

```
The dog bites the man.
```

 is different from:

```
The man bites the dog.
```

 Position information helps the model distinguish them.

 ### Segment Embedding

 Used in BERT to distinguish between two input segments, especially for tasks involving sentence pairs.

---

 # 12\. Step 4: Self-Attention

 This is one of the most important parts of BERT.

 Suppose BERT receives:

```
The bank approved the loan.
```

 When creating the representation of `bank`, self-attention allows it to consider other tokens:

```
The
bank       ← current token
approved
the
loan
```

 The model learns which tokens are important to one another.

 The basic attention equation is:

 $$
Attention(Q,K,V)
=
softmax\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
$$

 where:

 - $Q$ = Query
- $K$ = Key
- $V$ = Value

---

 # 13\. Query, Key, and Value Intuition

 A simple intuition is:

 ### Query

 > **What information am I looking for?**

 ### Key

 > **What information do I contain?**

 ### Value

 > **What information should I provide if I am relevant?**

 For example, when processing `bank`, the model can compare its query against the keys of other tokens.

 It might learn that:

```
bank → approved     important
bank → loan         important
bank → the          less important
```

 The attention mechanism then combines information from relevant tokens.

---

 # 14\. Multi-Head Attention

 BERT does not use just one attention mechanism.

 It uses **multi-head self-attention**.

 Different attention heads can learn different types of relationships.

 Conceptually:

```
                 Input
                   │
       ┌───────────┼───────────┐
       ↓           ↓           ↓
   Head 1       Head 2       Head 3
       ↓           ↓           ↓
 syntax       semantic     relationship
       │           │           │
       └───────────┼───────────┘
                   ↓
             Combined output
```

 One head might learn relationships related to syntax, while another can capture semantic or long-distance relationships.

 The exact interpretation of individual attention heads is more complicated than this simplified picture, but the general idea is that multiple heads provide different learned projections of the relationships among tokens.

---

 # 15\. Step 5: Feed-Forward Network

 After self-attention, BERT applies a **position-wise feed-forward neural network** to the representations.

 Conceptually:

```
Self-Attention
      ↓
Feed-Forward Network
      ↓
Updated Representation
```

 The feed-forward layer further transforms the information collected through attention.

---

 # 16\. Residual Connections and Layer Normalization

 BERT's Transformer layers also use:

 - **Residual connections**
- **Layer normalization**
- **Feed-forward networks**
- **Multi-head self-attention**

 A simplified Transformer encoder layer looks like:

```
Input
  ↓
Multi-Head Self-Attention
  ↓
Add & Normalize
  ↓
Feed-Forward Network
  ↓
Add & Normalize
  ↓
Output
```

 BERT stacks many such Transformer encoder layers.

 As information passes through these layers, the token representations become increasingly contextualized.

---

 # 17. How BERT Creates a Contextualized Embedding

 Consider:

```
The bank approved the loan.
```

 Initially, `bank` has a token representation.

 After passing through Transformer layers:

```
Layer 1
  ↓
bank + nearby relationships

Layer 2
  ↓
bank + stronger syntactic relationships

Layer 3
  ↓
bank + semantic relationships

...
  ↓
Final BERT representation
  ↓
bank in financial context
```

 Now consider:

```
The river bank was flooded.
```

 The same token `bank` passes through the network with different surrounding words:

```
river
flooded
```

 Therefore, its final representation changes.

 This is the core of contextualized embeddings.

---

 # 18\. BERT Pretraining

 BERT became particularly powerful because it was **pretrained on large amounts of text** before being adapted to specific NLP tasks.

 The original BERT training involved two major objectives:

 1. **Masked Language Modeling (MLM)**
2. **Next Sentence Prediction (NSP)**

---

 # 19\. Masked Language Modeling

 This is one of BERT's most important ideas.

 BERT randomly masks some tokens in a sentence.

 For example:

```
The bank approved the [MASK].
```

 BERT must predict:

```
loan
```

 Another example:

```
The cat is [MASK] on the mat.
```

 The model might predict:

```
sitting
```

 Because BERT is bidirectional, it can use information from both sides of the masked word.

 For example:

```
The bank approved the [MASK].
                         ↑
                  left context
The bank approved the

                         ↓
                  right context
```

 This forces the model to learn relationships between words and their surrounding context.

---

 # 20\. Why Masked Language Modeling Is Powerful

 Consider:

```
The animal crossed the road because it was tired.
```

 To understand the missing word in a masked version, the model must learn relationships among multiple words.

 For example:

```
The animal crossed the road because it was [MASK].
```

 Possible prediction:

```
tired
```

 The model learns patterns involving:

 - Syntax
- Semantics
- Word relationships
- Sentence structure
- Context

 Thus, BERT is not simply memorizing word frequencies.

 It is learning **representations that encode relationships among tokens in context**.

---

 # 21\. Next Sentence Prediction

 The original BERT also used **Next Sentence Prediction (NSP)**.

 The model receives two sentences:

```
Sentence A
Sentence B
```

 and predicts whether B logically follows A in the training data.

 For example:

```
A: I went to the store.
B: I bought some milk.
```

 The model may classify them as:

```
IsNext
```

 Whereas:

```
A: I went to the store.
B: The sun is a star.
```

 may be classified as:

```
NotNext
```

 NSP was intended to help BERT learn relationships between sentences. Later research found that NSP is not necessarily essential for all BERT-style pretraining approaches.

---

 # 22\. BERT Fine-Tuning

 After pretraining, BERT can be adapted to specific NLP tasks.

 This is called **fine-tuning**.

 For example:

```
             Pretrained BERT
                    ↓
          ┌─────────┼─────────┐
          ↓         ↓         ↓
     Sentiment    NER       QA
     Analysis
```

---

 # 23\. BERT for Sentiment Analysis

 Input:

```
[CLS] This movie was excellent! [SEP]
```

 The representation of `[CLS]` can be passed to a classification layer.

```
BERT
 ↓
[CLS] representation
 ↓
Classifier
 ↓
Positive
```

---

 # 24\. BERT for Named Entity Recognition

 Consider:

```
Barack Obama was born in Hawaii.
```

 BERT can produce contextualized representations for each token.

 A classification layer can then predict:

```
Barack → PERSON
Obama  → PERSON
Hawaii → LOCATION
```

---

 # 25\. BERT for Question Answering

 Given:

```
Context:
Paris is the capital of France.

Question:
What is the capital of France?
```

 BERT can process both the question and context:

```
[CLS] Question [SEP] Context [SEP]
```

 It can predict the location of the answer within the context:

```
Paris
```

---

 # 26\. BERT's Position in the NLP Timeline

 The overall development can now be shown as:

```
RULE-BASED NLP
       ↓
FREQUENCY-BASED NLP
       ↓
N-GRAMS / TF-IDF
       ↓
WORD2VEC / GLOVE
       ↓
STATIC WORD EMBEDDINGS
       ↓
ELMo
       ↓
CONTEXTUALIZED EMBEDDINGS
       ↓
TRANSFORMERS
       ↓
BERT
       ↓
Modern Transformer Models
       ↓
Large Language Models
```

 More conceptually:

```
COUNTING
   ↓
CO-OCCURRENCE
   ↓
REPRESENTATION
   ↓
CONTEXT
   ↓
ATTENTION
   ↓
UNDERSTANDING / GENERATION
```

---

 # 27\. Word2Vec vs. BERT

 | Feature | Word2Vec | BERT |
| --- | --- | --- |
| Representation | Static | Contextual |
| Vector for same word | Essentially fixed | Changes with context |
| Main architecture | Shallow neural model | Transformer encoder |
| Context | Local training window | Bidirectional sequence context |
| Attention | No | Yes |
| Handles polysemy | Limited | Much better |
| Training objective | Predict words from context / context from word | Masked Language Modeling + originally NSP |
| Sentence-level understanding | Limited | Stronger |
| Fine-tuning | Not the original paradigm | Central to BERT's use |
| Output | Word embeddings | Contextual token representations |

---

 # 28\. The Core Difference

 The difference can be summarized with one example.

 ### Word2Vec

```
bank
 ↓
one learned vector
```

 ### BERT

```
"The bank approved the loan."
             ↓
        contextual vector A

"The river bank was flooded."
             ↓
        contextual vector B
```

 Therefore:

 $$
\boxed{\text{Word2Vec: } word \rightarrow vector}
$$

 $$
\boxed{\text{BERT: } word + context \rightarrow contextual\ vector}
$$

---

 # 29\. Why BERT Was a Major Breakthrough

 BERT demonstrated that a large pretrained Transformer could learn **general-purpose contextual language representations** and then be adapted to many downstream NLP tasks.

 Instead of building a completely separate model from scratch for every task, the process became:

```
Large Corpus
     ↓
Pretraining
     ↓
General Language Representation
     ↓
Fine-Tuning
     ↓
Specific NLP Task
```

 This paradigm became extremely influential.

---

 # 30\. Final Conceptual Picture

 The evolution from frequency to contextualized embeddings can be summarized as:

```
                    NLP EVOLUTION

Frequency
   │
   │ "How often does it occur?"
   ↓
N-Grams / TF-IDF
   │
   │ "What words occur together?"
   ↓
Word2Vec / GloVe
   │
   │ "What semantic relationships
   │  exist between words?"
   ↓
Static Word Embeddings
   │
   │ One word → one vector
   ↓
Contextualized Embeddings
   │
   │ Word + context → vector
   ↓
Attention
   │
   │ "Which parts of the context matter?"
   ↓
Transformer
   │
   │ Self-attention across tokens
   ↓
BERT
   │
   │ Bidirectional contextual representation
   │ + Masked Language Modeling
   ↓
Modern NLP / LLMs
```

 ## Key Takeaway

 > **Word2Vec was a major bridge from frequency-based NLP to representation learning: it learned semantic vectors from word co-occurrence. BERT took the next major step by making those representations contextual—so the same word could have a different vector depending on the words surrounding it.**

 In the simplest form:

 $$
\boxed{
\text{Frequency}
\rightarrow
\text{Word2Vec}
\rightarrow
\text{Context}
\rightarrow
\text{Attention}
\rightarrow
\text{BERT/Transformers}
}
$$

 And the fundamental change is:

 > **From "What is this word?" to "What does this word mean in this context?"**

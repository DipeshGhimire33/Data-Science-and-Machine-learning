## Gender and Occupation

```python
df.groupby("gender")["occupation"].value_counts().plot(kind = "barh")
```

The data show that majourity of females were doing " Administrative Clerical" work whereas majourity of males were working "Craft-Repairs".

___

## Income , Occupation, and Gender

```python
plt.figure(figsize=(20,30))
df.groupby("income")[["occupation","gender"]].value_counts().plot(kind = "barh")
```

The figure highlights significant gender income disparities, with males dominating high-income roles and females more concentrated in lower-income roles. Occupations such as executive managerial and professional specialties are associated with higher earnings, while roles like farming, fishing, and protective services tend to have lower income levels, especially among females.

___

## Income, Education and Gender

```python
plt.figure(figsize=(20,30))
df.groupby("income")[["education","gender"]].value_counts().plot(kind = "barh")
```

The chart underscores a significant gender income gap, with males generally earning more than females at similar levels of education. Higher educational attainment correlates with increased income for both genders, but disparities remain, particularly at the higher end of the income spectrum. The data emphasizes the importance of education in achieving higher income levels and highlights persistent gender-based income differences across occupational and educational categories.

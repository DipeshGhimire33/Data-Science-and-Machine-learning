# Stastical analysis to verify Chart analysis

```python
from scipy.stats import chi2_contingency

categorical_features = [
    'gender',
    'stress_level',
    'academic_work_impact'
]

for feature in categorical_features:
    # Create contingency table
    table = pd.crosstab(df[feature], df['addicted_label'])

    # Chi-square test
    chi2, p, dof, expected = chi2_contingency(table)

    # Cramér's V
    n = table.sum().sum()
    min_dim = min(table.shape) - 1
    cramers_v = (chi2 / (n * min_dim)) ** 0.5

    print(f"\n{'-'*50}")
    print(f"Feature: {feature}")
    print(f"{'-'*50}")
    print(table)
    print(f"\nChi-square statistic: {chi2:.4f}")
    print(f"P-value: {p:.6f}")
    print(f"Cramér's V: {cramers_v:.4f}")
```

___

Overall, the bar chart and statistical analysis show that **gender, stress level, and academic work impact have very similar distributions of the target variable `addicted_label` across their categories**. Although the Chi-square tests indicate statistically significant associations, the **Cramér’s V values are extremely small**, indicating that these associations are **negligible to very weak in practical terms**. Therefore, these categorical variables do not appear to have a strong individual relationship with addiction status, despite the statistical significance resulting from the large sample size.

## Notes :

Statistical analysis is necessary because a **bar chart only shows what the data looks like**. It doesn't tell us whether the observed difference is likely to be real or could simply be due to random variation.

For Our project, we can think of statistical analysis as answering:

> **“Is there actually a relationship between this feature and my target, and if so, how strong is it?”**

### Common statistical analyses

| Situation                   | Variables                         | Useful analysis                | What it tells you                                           |
| --------------------------- | --------------------------------- | ------------------------------ | ----------------------------------------------------------- |
| Categorical ↔ Categorical   | Gender ↔ Addiction                | **Chi-square test**            | Is there an association?                                    |
| Categorical ↔ Categorical   | Gender ↔ Addiction                | **Cramér's V**                 | How strong is the association?                              |
| Categorical ↔ Numeric       | Gender ↔ Salary                   | **t-test** (2 groups)          | Do two group means differ?                                  |
| Categorical ↔ Numeric       | Department ↔ Salary               | **ANOVA** (3+ groups)          | Do group means differ?                                      |
| Categorical ↔ Numeric       | Department ↔ Salary               | **Kruskal-Wallis**             | Same idea as ANOVA, without assuming normality              |
| Numeric ↔ Numeric           | Age ↔ Income                      | **Pearson correlation**        | Strength/direction of linear relationship                   |
| Numeric ↔ Numeric           | Age ↔ Income                      | **Spearman correlation**       | Monotonic relationship; useful with non-normal/ordinal data |
| Numeric ↔ Binary target     | Age ↔ Addiction (0/1)             | **Point-biserial correlation** | Relationship between numeric feature and binary target      |
| Numeric → Numeric           | Hours studied → Score             | **Linear regression**          | Predict/estimate a numeric target                           |
| Multiple features → Numeric | Age + hours + etc. → Salary       | **Multiple linear regression** | Effect/prediction using several variables                   |
| Multiple features → Binary  | Age + gender + stress → Addiction | **Logistic regression**        | Predict probability of a binary outcome                     |

### For your particular dataset

Our target is:

**`addicted_label` = 0 or 1**

So the analysis depends on the type of feature we're examining.

#### 1. Categorical feature → binary target

For example:

`gender → addicted_label`

Use:

**Chi-square + Cramér's V**

* Chi-square → *Is there statistically significant association?*
* Cramér's V → *How strong is the association?*

That's exactly what you did.

#### 2. Numeric feature → binary target

For example:

`screen_time → addicted_label`

We could use:

* **Point-biserial correlation**
* **Mann-Whitney U test**
* Or compare distributions with **box plots**

#### 3. Numeric → numeric

For example:

`screen_time → sleep_hours`

Use:

* **Pearson correlation** if the relationship is approximately linear
* **Spearman correlation** if the relationship is monotonic but not necessarily linear

### Most important thing

Don't rely on **p-value alone**.

A good statistical analysis usually considers:

**Visualization → Statistical significance → Effect size → Practical interpretation**

For current example:

> Bar chart → categories look similar
> Chi-square → statistically significant
> Cramér's V → extremely weak association
> Final interpretation → **statistically detectable, but practically negligible relationship**

That's a much stronger analysis than simply saying *“p < 0.05, therefore there is a strong relationship.”*

Performing taining and testing for using same training dataset.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
```

```python
X = imputed_df.drop(columns = "addicted_label")
y = imputed_df[["addicted_label"]]
```

```python
X_train , X_test ,y_train ,y_test = train_test_split(X, y ,test_size= 0.2, random_state= 999, stratify= y)
```

```python
scaler =StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
model = LogisticRegression().fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_test , y_pred)
```

```python
confusion_matrix(y_test , y_pred)
```

```python
print(classification_report(y_test , y_pred))
```

```python
corr = df.corr()
plt.figure(figsize=(10,10))
sns.heatmap(corr,annot= True)
```

```python
df[["alcohol","quality"]].corr(numeric_only= True)
```

* Fixed acidity shows a positive correlation with citric acid (0.67), suggesting they tend to increase together.
* pH has a negative correlation with fixed acidity (-0.68) and alcohol (-0.2), indicating that higher acidity corresponds to lower pH (more acidic) and slightly less alcohol.
* Residual sugar has a weak positive correlation with density (0.2), which makes sense since higher sugar content increases density.
* Several variables like volatile acidity, residual sugar, and sulphates have weak correlations with quality, indicating they may not strongly influence overall quality independently.

Overall, the most notable relationship is between alcohol and quality, suggesting that alcohol content is a significant factor in the perceived quality of the samples.
